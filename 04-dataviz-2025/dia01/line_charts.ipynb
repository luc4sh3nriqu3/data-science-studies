```python
import matplotlib.pyplot as plt
import seaborn as sns
```

Vamos imaginar que este seja um gráfico contendo informações de coletas de maçãs (em toneladas) coletadas de uma safra ao longo de 6 anos. Os dados estão dispostos abaixo:

```python
yield_apples = [0.895, 0.91, 0.919, 0.926, 0.929, 0.931]
```

```python
plt.plot(yield_apples)
```

```python
#usando ; ao final da linha para evitar a exibição do objeto retornado pela função plt.plot();
plt.plot(yield_apples);
```

Note que o eixo x não mostra qual o respectivo ano daquela coleta. Vamos manipular essa dimensão para acrescentar os dados de anos.

```python
years = [2015, 2016, 2017, 2018, 2019, 2020]
yield_apples = [0.895, 0.91, 0.919, 0.926, 0.929, 0.931]
plt.plot(years, yield_apples);
```

## Legendas nos eixos

Podemos adicionar labels nos eixos x e y para informar o que cada um deles representa utilizando os métodos `plt.xlabel` e `plt.ylabel`

NOTA: a ordem dos comandos (plot, definição de eixos, etc) não importa desde que estejam na mesma célula do jupyter, tudo o que estvier definido naquela célula será aplicada a um mesmo plot

```python
plt.plot(years, yield_apples)
plt.xlabel('Year')
plt.ylabel('Yield (tons per hectare)')
```

## Plotando múltiplas linhas

Para plotar múltiplas linhas, podemos definir os dados que cada uma terá e a única coisa que precisamos fazer é invocar `plt.plot` de todos os dados que qeueremos desenhar na mesma célula

```python
years = range(2000, 2012)
apples = [0.895, 0.91, 0.919, 0.926, 0.929, 0.931, 0.934, 0.936, 0.937, 0.9375, 0.9372, 0.939]
oranges = [0.962, 0.941, 0.930, 0.923, 0.918, 0.908, 0.907, 0.904, 0.901, 0.898, 0.9, 0.896]
```

```python
plt.plot(years, apples)
plt.plot(years, oranges)
plt.xlabel('Year')
plt.ylabel('Yield (tons per hectare)');
```

## Título e Lgenda

Podemos adicionar títulos ao gráfico através da função `plt.title` e legendas a pelo comando `plt.legend` (as legendas serão atribuídas na mesma sequencia que os pltos, ou seja, se pltamos primeiro as maçãs e depois as laranjas, a lista de legendas ficará primeiro as maçãs e depois as laranjas ['apples', 'oranges'])

```python
#Definindo os dados que iremos plotr
plt.plot(years, apples)
plt.plot(years, oranges)

#Definindo os rótulos dos eixos
plt.xlabel('Years')
plt.ylabel('Yield (tons per hectare)')

#Adicionando título e legenda ao gráfico
plt.title('Apple and Orange Yield Over Years')
plt.legend(['Apples', 'Oranges']);
```

## Marcadores de Linha

Podemos adicionar marcadores às linhas através do argumento `marker` na função `plt.plot`. Os marcadores podem ser pontos, linhas, quadrados, etc. Podemos conferir a lista completa de marcadores aqui:
https://matplotlib.org/3.1.1/api/markers_api.html

```python
#Definindo os dados que iremos plotr
plt.plot(years, apples, marker='o')
plt.plot(years, oranges, marker='x')

#Definindo os rótulos dos eixos
plt.xlabel('Years')
plt.ylabel('Yield (tons per hectare)')

#Adicionando título e legenda ao gráfico
plt.title('Apple and Orange Yield Over Years')
plt.legend(['Apples', 'Oranges']);
```

## Estilização e marcadores

A função plt.plot suporta diversos argumentos para estilização e marcação:<br>
- `color` ou `c`: define a cor da linha (cores suportadas: https://matplotlib.org/3.1.0/gallery/color/named_colors.html)
- `linestyle` ou `ls`: escolhe entre sólida ('solid') ou tracejada ('dashed')
- `linewidth` ou `lw`: define a espessura da linha
- `markersize` ou `ms`: define o tamanho dos marcadores
- `markeredgecolor` ou `mec`: define a cor da borda para marcadores
- `markeredgewidth` ou `mew`: define a espessura da borda para marcadores
- `markerfacecolor` ou `mfc`: define a cor de preenchimento para marcadores
- `alpha`: define a opacidade do plot
<br>
Podemos checar toda a documentação do `plt.plot` aqui: <br>
https://matplotlib.org/api/_as_gen/matplotlib.pyplot.plot.html#matplotlib.pyplot.plot

```python
#Definindo os dados que iremos plotr
plt.plot(years, apples, marker='s', c='b', ls='-', lw=2, ms=8, mew=2, mec='navy')
plt.plot(years, oranges, marker='o', c='r', ls='--', lw=3, ms=10, alpha=.5)

#Definindo os rótulos dos eixos
plt.xlabel('Years')
plt.ylabel('Yield (tons per hectare)')

#Adicionando título e legenda ao gráfico
plt.title('Apple and Orange Yield Over Years')
plt.legend(['Apples', 'Oranges']);
```

`fmt` é o argumento para encurtar especificações de linhas que pode especificar o estilo da linha, o marcador e a cor. Ele pode ser declarado como terceiro argumento da função `plt.plot`.<br>
fmt = '[marker][line][color]'

```python
#Definindo os dados que iremos plotr
plt.plot(years, apples, 's-b')
plt.plot(years, oranges, 'o--r')

#Definindo os rótulos dos eixos
plt.xlabel('Years')
plt.ylabel('Yield (tons per hectare)')

#Adicionando título e legenda ao gráfico
plt.title('Apple and Orange Yield Over Years')
plt.legend(['Apples', 'Oranges']);
```

Se nenhuma estilização de linha for definida no `fmt`, somente os marcadores serão desenhados (a linha sumirá)

```python
plt.plot(years, oranges, 'or')
plt.title('Orange Yield Over Years')
```

## Mudando o tamanho da figura (plot)
podemos usar a função `plt.figure` para mudar o tamanho da figura

```python
plt.figure(figsize=(12, 6))
plt.plot(years, oranges, 'or')
plt.title('Orange Yield Over Years')
```

## Usando Seaborn para utilizar estilizações padrões pré-definidas

Uma maneira rápida e fácil de melhorarmos nossos gráficos é usando alguma estilização padrão definida na biblioteca do Seaborn. Elas podem ser aplicadas globalmente usando a função `sns.set_style`. Podemos ver a lista completa de estilizações pré-definidas aqui: https://seaborn.pydata.org/generated/seaborn.set_style.html

```python
sns.set_style('whitegrid')  # Exemplos de estilos: darkgrid, whitegrid, dark, white, ticks
```

```python
#Definindo os dados que iremos plotr
plt.plot(years, apples, 's-b')
plt.plot(years, oranges, 'o--r')

#Definindo os rótulos dos eixos
plt.xlabel('Years')
plt.ylabel('Yield (tons per hectare)')

#Adicionando título e legenda ao gráfico
plt.title('Apple and Orange Yield Over Years')
plt.legend(['Apples', 'Oranges']);
```

```python
sns.set_style('darkgrid')
```

```python
#Definindo os dados que iremos plotr
plt.plot(years, apples, 's-b')
plt.plot(years, oranges, 'o--r')

#Definindo os rótulos dos eixos
plt.xlabel('Years')
plt.ylabel('Yield (tons per hectare)')

#Adicionando título e legenda ao gráfico
plt.title('Apple and Orange Yield Over Years')
plt.legend(['Apples', 'Oranges']);
```

Precisamos definir o `sns.set_style` uma única vez e a partir daí os gráficos subsequentes terão aquela estilização, como mostrado abaixo:

```python
#Perceba que não utilizamos o sns.set_style aqui, mas mesmo assim o estilo aplicado anteriormente permanece
plt.plot(years, apples, 'or')
plt.title('Apple Yield Over Years')
```

Podemos também editar usando as estilizações padrão usando o dicionário `matplotlib.rcParams`.<br> Podemos ver mais sobre aqui: https://matplotlib.org/3.2.1/tutorials/introductory/customizing.html#matplotlib-rcparams
