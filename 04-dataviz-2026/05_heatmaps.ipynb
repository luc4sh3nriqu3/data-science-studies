# Heatmap

Heatmaps são utilizados para representar dados bidimensionais (matrizes ou tabelas) utilizando cores.
Vamos entender melhor utilizando um exemplo

```python
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib
```

O dataset que iremos usar mostra a quantidade de passageiros que passaram em um aeroport mês a mês durante um período de anos. Usaremos o heatmap para analisar o sentido do fluxo (se a quantidade aumentou ou diminuiu). Para isso, precisaremos transformar nossos dados em uma tabela (usando pivot)

```python
flights_df = sns.load_dataset('flights').pivot(index='month', columns='year', values='passengers')
```

```python
flights_df
```

Aqui podemos ver a quantidade de passageiros que passaram pelo aeroporto (x1000) ao decorrer dos meses de um ano. Utilizaremos a função `sns.heatmap` para visualizar os dados.

```python
sns.set_style('darkgrid')
matplotlib.rcParams['figure.figsize'] = (9, 5)
```

```python
plt.title('Número de passageiros (1000s)')
#sns.heatmap(flights_df);
```

Aqui, as cores mais escuras representam um fluxo menor de passageiros enquanto as cores mais claras indicam um fluxo maior naquele mês. E analisando o gráfico podemos concluir que:
- O fluxo maior de passageiros é por volta de Jul e Ago
- O fluxo aumentou conforme os anos foram passando

```python
plt.title('Número de passageiros (1000s)')
sns.heatmap(flights_df, fmt='d', annot=True, cmap='Blues');
```
