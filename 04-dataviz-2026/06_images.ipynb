Com o matplotlib podemos trabalhar também com imagens. Vamos baixar uma imagem da internet

```python
from urllib.request import urlretrieve
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib
import numpy as np
from PIL import Image

sns.set_style('darkgrid')
matplotlib.rcParams['figure.figsize'] = (9, 5)
```

Aqui, para que o sistema não bloqueie o acesso (achando que somos bots ou pra limitar acessos repetidos através de scripts) precisaremos utilizar um user-agent para acessar a página e fazer a requisição

```python
import requests

url = 'https://i.imgur.com/SkPbq.jpg'

headers = {
    'User-Agent': 'Mozilla/5.0'
}

response = requests.get(url, headers=headers)

with open('chart.jpeg', 'wb') as f:
    f.write(response.content)
```

```python
from PIL import Image
```

```python
img = Image.open('chart.jpeg')
```

```python
import numpy as np
```

Uma imagem quando importada pelo PIL, vira uma matriz tridimensional numpy contendo as intensidades de cada tom RGB da imagem (cada cor é de uma camada). Nós podemos converter ela para um array utilizando `np.array`.

```python
img_array = np.array(img)
```

```python
img_array.shape
```

Nós podemos mostrar a imagem com o comando `plt.imshow`.

```python
plt.imshow(img);
```

Nós podemos retirar as representações dos eixos, retirar linhas de grade e colocar títulos em imagens conforme é mostrado a seguir

```python
plt.grid(False)
plt.title('Meme de Data Science')
plt.axis('off')
plt.imshow(img);
```

Se quisermos mostrar um pedaço da imagem, podemos selecionar a fatia pelo array numpy

```python
plt.grid(False)
plt.axis('off')
plt.title('Meme de Data Science')
plt.imshow(img_array[125:325, 105:305]);
```
