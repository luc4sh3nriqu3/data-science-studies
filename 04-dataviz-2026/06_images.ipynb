# Imagens
Com o matplotlib podemos trabalhar também com imagens. Vamos baixar uma imagem da internet

```python
from urllib.request import urlretrieve
import matplotlib.pyplot as plt
```

Aqui, para que o sistema não bloqueie o acesso (achando que somos bots ou pra limitar acessos repetidos através de scripts) precisaremos utilizar um user-agent para acessar a página e fazer a requisição

```python
import requests

url = 'https://i.imgur.com/SkPbq.jpg'

headers = {
    'User-Agent': 'Mozilla/5.0'
}

response = requests.get(url, headers=headers)

with open('chart.jpeg', 'wb') as f:
    f.write(response.content)
```

```python
from PIL import Image
```

```python
img = Image.open('chart.jpeg')
```

```python
import numpy as np
```

Uma imagem quando importada pelo PIL, vira uma matriz tridimensional numpy contendo as intensidades de cada tom RGB da imagem (cada cor é de uma camada). Nós podemos converter ela para um array utilizando `np.array`.

```python
img_array = np.array(img)
```
