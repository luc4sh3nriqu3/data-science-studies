# Gráficos de Barras

Eles são utilizados para comparar categorias de dados utilizando barras verticais, onde o comprimento ou altura de cada barra é proporcional ao valor que representa. Eles são semelhantes aos gráficos de linhas no sentido de representar uma informação de duas dimensões porém, os gráficos de linhas são mais utilizados para analisar tendências, os gráficos de barras são melhores para comparações entre os próprios dados (comparar barra a barra, que não é tão fácil utilizando os gráficos de linnhas)

```python
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib
```

Nós podemos utilizar a função `plt.bar` para desenhar um gráfico de barras

```python
years = range(2000, 2006)
apples = [0.35, 0.6, 0.9, 0.8, 0.65, 0.8]
oranges = [0.4, 0.8, 0.9, 0.7, 0.6, 0.8]
```

```python
plt.rcParams['figure.figsize'] = (9, 5)
sns.set_style('darkgrid')
```

```python
plt.plot(years, oranges);
```

```python
plt.bar(years, oranges);
```

Percebe-se uma semelhança entre os gráficos, porém com a ressalva de que nos gráficos de barra podemos comparar melhor os dados enquanto no gráfico de linhas podemos analisar melhor as tendências. Uma boa alternativa é utilizar os dois em conjunto.

```python
plt.plot(years, apples, 'o--r')
plt.bar(years, apples)
plt.title('Yield of Apples')
```

Assim como os histogramas, podemos escrever gráficos de barrasem conjunto. Faremos isso manipulando o parâmetro `bottom` em `plt.bar`

```python
plt.bar(years, apples)
plt.bar(years, oranges, bottom=apples);
# Bottom = apples faz com que as barras de maçãs sejam representadas abaixo
```

Aqui a visualização fica o seguinte: Cada barra será representada pela soma de cada uma das classes naquele período, ou seja: em 2000 temos maçãs = 0.35 e laranjas = 0.4 -> 0.35 + 0.4 = 0.75

## Gráficos de barras com intervalos

Agora veremos um exemplo utilizando o dataset incluso no seaborn que contém informações de sexo, data de nascimento, valor da conta e gorjetas de um restaurante, durante o período de uma semana

```python
tips_df = sns.load_dataset("tips")
```

```python
tips_df
```

Aqui, poderiamos saber quanto que os clientes gastaram em média a cada dia da semana para comparar entre os dias, para isso precisaremos fazer uma espécie de GROUP BY. Esse primeiro exemplo foi feito sem o uso do seaborn (como exercício de treinamento)

```python
tips_a_day = tips_df.groupby('day')['total_bill'].mean() #Agrupando por dia e calculando a média da conta total
plt.bar(tips_a_day.index, tips_a_day.values);
print(tips_a_day)
```

Agora faremos utilizando a biblioteca do seaborn que já possui essa função acoplada, evitando escrever mais linhas de código (pensando em casos onde precisariamos gerar vários gráficos de agrupamento)

```python
sns.barplot(x='day', y='total_bill', data=tips_df);
```

Outra coisa que esse plot do seaborn trás, é a informação de variação em relação à média mostrada (representada pelas linhas nos topos das barras)

Podemos adicionar também o marcador através do parâmetro `hue` para podermos por exemplo ver a média de gastos comparando homens e mulheres para esses dias.

```python
sns.barplot(x='day', y='total_bill', hue='sex', data=tips_df);
```

Aqui, podemos também inverter os eixos x e y para deixarmos as barras na horizontal

```python
sns.barplot(x='total_bill', y='day', hue='sex', data=tips_df);
```
