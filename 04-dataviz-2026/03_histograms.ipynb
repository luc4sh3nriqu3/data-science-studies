```python
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib
```

# Histogramas

Histogramas são bons para analisar distribuição de dados, pois ele separa os dados em intervalos de valores e agrupa (conta) os dados que estão no mesmo intervalo. (Eixo x -> Valores) (Eixo y -> Contagem)

```python
flowers_df = sns.load_dataset('iris')
```

```python
flowers_df.describe()
```

Podemos plotar u histograma através da função `plt.hist`. Como mostrado abaixo

```python
sns.set_style('darkgrid')
plt.rcParams['figure.figsize'] = (9, 5)
```

```python
plt.title('Histograma da largura das pétalas (Sepal Width)')
plt.hist(flowers_df['sepal_width']);
```

Aqui nós podemos ver que os dados:
- Estão no intervalo [2.0 - 4.5]
- A maioria das larguras vão de [3.0, 3.5] (Onde estão a maior quantidade de dados -> +60 amostras)

## Controlando os intervalos

Podemos controlar os intervlos definidos no histograma pelo parâmetro `bins`

```python
import numpy as np
```

```python
plt.hist(flowers_df['sepal_width'], bins=5);
```

```python
#Cria os bins personalizados, definindo o intervalo entre 2 e 5 com passo de 0.25
np.arange(2, 5, 0.25)
```

```python
plt.hist(flowers_df['sepal_width'], bins=np.arange(2, 5, 0.25));
```

```python
#Definindo bins de diferentes tamanhos
plt.hist(flowers_df['sepal_width'], bins=[1, 3, 4, 4.5]);
```

## Multiplos Histogramas

Vamos desenhar histogramas para cada espécie de flores

```python
setosa_df = flowers_df[flowers_df['species'] == 'setosa']
versicolor_df = flowers_df[flowers_df['species'] == 'versicolor']
virginica_df = flowers_df[flowers_df['species'] == 'virginica']
```

```python
plt.hist(setosa_df['sepal_width'], alpha=0.4, bins=np.arange(2, 5, 0.25))
plt.hist(versicolor_df['sepal_width'], alpha=0.4, bins=np.arange(2, 5, 0.25))

plt.legend(['Setosa', 'Versicolor']);
```

Podemos plotar também multiplos histogramas um sobre o outro

```python
plt.hist([setosa_df['sepal_width'], versicolor_df['sepal_width'], virginica_df['sepal_width']],
         bins=np.arange(2, 5, 0.25),
         stacked=True,);

plt.legend(['Setosa', 'Versicolor', 'Virginica']);
```

Aqui podemos analisar a distribuição conjunta das três espécies tem suas larguras definidas em +-3.0 e vão até +-3.5
