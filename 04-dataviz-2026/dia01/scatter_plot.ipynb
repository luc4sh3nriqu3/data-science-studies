# Scatter Plot

Sccater Plot são gráficos utilizados para plotar pontos com 2 dimensões. Podemos também adicionar cor ao ponto para utilizar como terceira dimensão

```python
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib
```

```python
flowers_df = sns.load_dataset('iris')
```

```python
flowers_df
```

```python
flowers_df['species'].unique()
```

Vamos agora tentar analisar a relação entre `sepal_width` e `sepal_length`. Primeiro vamos tentar utilizar o gráfico de linhas através da `plt.plot`. Porém, perceberemos que não conseguiremos absorver muita informação da saída do gráfico pois há muitas combinações entre essas duas variáveis.

```python
sns.set_style('darkgrid')
matplotlib.rcParams['figure.figsize'] = (9, 5)
```

```python
plt.plot(flowers_df['sepal_length'], flowers_df['sepal_width'])
```

Agora, mostraremos os mesmos dados, porém utilizando Scatter Plot (Gráficos de Dispersão). Perceberemos que o resultado será muito melhor. O comando para plotar gráficos de dispersão é: `sns.scatterplot`

```python
sns.scatterplot(x=flowers_df['sepal_length'], y=flowers_df['sepal_width']);
```

Aqui podemos notar que existem aglomerados (clusters) de flores com valores de relação entre as duas variáveis parecidos, porém não podemos ver esses padrões apenas plotando o gráfico normalmente. Para nos ajudar usaresmos o parâmetro `hue`. Poderemos também aumentar o tamanho dos pontos usando `s`

```python
sns.scatterplot(x=flowers_df['sepal_length'], 
                y=flowers_df['sepal_width'],
                hue=flowers_df['species'], 
                s=100);
```

Aqui as cores de cada classe assumirão a mesma cor. Assim, fica muito melhor a visualização

## Customizando Plots do Seaborn

Ainda que as duas bibliotecas possuem suas funções de plot independentes e internamente, nós podemos combinar elas para um plot final.

```python
plt.figure(figsize=(12, 6))
plt.title('Sepal Dimensions')

sns.scatterplot(x=flowers_df['sepal_length'], 
                y=flowers_df['sepal_width'],
                hue=flowers_df['species'], 
                s=100);
```

Outra maneira de importar dados do pandas para plotar é utilizando o parametro `data`. Assim, não precisamos passar uma coluna (lista) do dataset sempre que formos plotar, mas passamos apenas os nomes.

```python
plt.title('Sepal Dimensions')
sns.scatterplot(x='sepal_length', 
                y='sepal_width',
                hue='species', 
                s=100,
                data=flowers_df);
```
